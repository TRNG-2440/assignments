```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("insurance.csv")
df
```

```python
print(df.dtypes)
print(df.isna().sum())
```

```python
### Question 1 — Do females or males typically have higher medical costs?

df_female_med_costs = df[df["sex"] == "female"]
df_male_med_costs = df[df["sex"] == "male"]
print(f"Female Mean Medical Costs: {df_female_med_costs['charges'].mean():.02f}")
print(f"Male Mean Medical Costs: {df_male_med_costs['charges'].mean():.02f}")
print(f"Female Median Medical Costs: {df_female_med_costs['charges'].median():.02f}")
print(f"Male Median Medical Costs: {df_male_med_costs['charges'].median():.02f}")

fig, ax = plt.subplots()
fig.suptitle("Females vs Males on Med Costs")
ax.hist(df_female_med_costs["charges"], bins=30, alpha=0.5, label="Female")
ax.hist(df_male_med_costs["charges"], bins=30, alpha=0.5, label="Male")

plt.axvline(df_female_med_costs['charges'].mean(), linestyle="--", alpha=0.5, color="red", label=f"Mean: {df_female_med_costs['charges'].mean():.0f}")
plt.axvline(df_male_med_costs['charges'].mean(), linestyle="-", alpha=0.5, color="black", label=f"Mean: {df_male_med_costs['charges'].mean():.0f}")
plt.axvline(df_female_med_costs['charges'].median(), linestyle="--", alpha=0.5, color="red", label=f"Median: {df_female_med_costs['charges'].median():.0f}")
plt.axvline(df_male_med_costs['charges'].median(), linestyle="-", alpha=0.5, color="black", label=f"Median: {df_male_med_costs['charges'].median():.0f}")
plt.show()
```
