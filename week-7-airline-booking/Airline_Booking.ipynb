## Question 28 — Numeric functions
Create fare_plus_tax, fare_after_flat_discount, doubled_baggage, and fare_per_kg using arithmetic operators. Use a safe divisor for zero baggage.

**Functions/concepts:** `arithmetic operators, when`

```python
display(
    bookings_origin_dest_code_df.select(
        (F.col('ticket_amount') + F.lit(2.99)).alias('fare_plus_tax'),
        (F.col('ticket_amount') - F.lit(2.99)).alias('fare_after_flat_discount'),
        (F.col('baggage_kg') * F.lit(2)).alias('doubled_baggage'),
        F.when(F.col('baggage_kg') != 0, F.col('ticket_amount') / F.col('baggage_kg')).otherwise(F.lit(0)).alias('fare_per_kg')
    )
)
```

## Question 29 — Conditional logic
Create fare_category: PREMIUM for fare >= 10,000; STANDARD for fare >= 6,000; otherwise BUDGET.

**Functions/concepts:** `when, otherwise`

```python
bookings_fare_category_df = bookings_origin_dest_code_df.withColumn(
    "fare_category",
    F.when(F.col("ticket_amount") >= 10000, F.lit("PREMIUM"))
    .when(F.col("ticket_amount") >= 6000, F.lit("STANDARD"))
    .otherwise(F.lit("BUDGET")),
)
display(bookings_fare_category_df)
```

## Question 66 — Partitions
Reduce sql_airline_aggregation_df to one partition using DataFrame.coalesce(1). Explain why this should not be used blindly for large production data.

**Functions/concepts:** `DataFrame.coalesce`

```python
print(f"Records: {sql_airline_aggregation_df.coalesce(1).collect()}")
```

`DataFrame.coalesce(1)` should be used carefully, since it reduces the number of partitions, and thus the number of executors to 1. If the data is too large, OOM errors can occur. It also affects processing speed negatively, since only one executor is doing all the processing.
